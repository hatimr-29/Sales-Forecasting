WEEK 5 FEATURE ENGINEERING

```python
df = pd.read_csv("sales_data_cleaned.csv")
df
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
# Year
df["Year"] = df["Order Date"].dt.year

# Month
df["Month"] = df["Order Date"].dt.month

# Day
df["Day"] = df["Order Date"].dt.day

# Quarter
df["Quarter"] = df["Order Date"].dt.quarter

# Month Name
df["Month Name"] = df["Order Date"].dt.strftime("%B")

# Day of Week
df["Day of Week"] = df["Order Date"].dt.day_name()
```

```python
df["Is Weekend"] = np.where(
    df["Order Date"].dt.dayofweek >= 5,
    1,
    0
)
```

```python
df["Avg Price"] = df["Total Sales"] / df["Quantity"]
```

```python
bins = [0, 1000, 5000, 20000, np.inf]

labels = ["Low", "Medium", "High", "Premium"]

df["Price Category"] = pd.cut(
    df["Unit Price"],
    bins=bins,
    labels=labels
)
```

```python
print(df.columns.tolist())
```

```python
df["Month Name"] = df["Order Date"].dt.strftime("%B")

df["Day of Week"] = df["Order Date"].dt.day_name()

bins = [0,1000,5000,20000,float("inf")]
labels = ["Low","Medium","High","Premium"]

df["Price Category"] = pd.cut(df["Unit Price"], bins=bins, labels=labels)
df
```

```python
cols = ["Category", "Region", "Month Name", "Day of Week", "Price Category"]

for col in cols:
    if col in df.columns:
        print(f"✓ {col} exists")
    else:
        print(f"✗ {col} NOT FOUND")
```

```python
df = pd.get_dummies(
    df,
    columns=[
        "Month Name",
        "Day of Week",
        "Price Category"
    ],
    drop_first=True
)
```

```python
features = [
    "Quantity",
    "Unit Price",
    "Year",
    "Month",
    "Day",
    "Quarter",
    "Is Weekend",
    "Avg Price"
]

X = df[features]

y = df["Total Sales"]
```

```python
training_data = pd.concat([X, y], axis=1)

training_data.head()
```

```python
training_data.to_csv(
    "sales_training_dataset.csv",
    index=False
)

print("Feature Engineering Completed Successfully!")
```

```python
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("sales_training_dataset.csv")
```

```python
X = df[[
    "Quantity",
    "Unit Price",
    "Year",
    "Month",
    "Quarter",
    "Is Weekend"
]]

y = df["Total Sales"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Random Forest Model Trained Successfully!")
```

```python
y_pred = model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("Model Performance")
print("-------------------")
print("MAE :", round(mae,2))
print("RMSE:", round(rmse,2))
print("R² Score:", round(r2,4))
```

```python
joblib.dump(model, "sales_forecasting_model.pkl")

print("Model Saved Successfully!")
```

```python
loaded_model = joblib.load("sales_forecasting_model.pkl")

print("Model Loaded Successfully!")
```

```python
sample = X_test.iloc[[0]]

prediction = loaded_model.predict(sample)

print("Predicted Sales:", prediction[0])
print("Actual Sales:", y_test.iloc[0])
```
